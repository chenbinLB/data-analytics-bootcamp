## 0. AI 时代的数据分析师

### 0.1 现状

大厂数分岗从 2024 起面试必问:
- "你怎么用 LLM 提升工作效率?"

### 0.2 不会用 AI 的数分 vs 会用的数分

| 维度 | 不会用 AI | 会用 AI |
|------|---------|---------|
| SQL 写法 | 写 30 分钟,自己试错 | AI 30 秒出框架,自己调整 |
| 业务分析 | 从 0 想,可能漏维度 | AI 头脑风暴出 10 个角度,挑 3 个深挖 |
| 数据清洗 | 手动写 20 行 pandas | AI 5 行搞定 + 解释 |

| 面试准备 | 找题自己做 | AI 出题 + 反馈,效率翻倍 |

### 0.3 大厂铁律

- **AI 是工具,不是替代**:AI 写 SQL,但你要 review 对不对
- **不要把业务问题直接抛给 AI**:AI 不知道你公司的具体业务
- **AI 输出的代码必须跑通**:跑不通 = 浪费时间
- **Prompt 工程是新基本功**:会写 prompt 的人,效率是普通人的 3-5 倍


## 1. Prompt 工程基础

### 1.1 原理

**Prompt 5 要素**:
1. **角色**:你让 AI 当什么(资深数分 / SQL 大神)
2. **任务**:你要它做什么(写 SQL / 解释数据)
3. **上下文**:背景信息(表结构 / 业务场景)
4. **约束**:格式 / 风格 / 限制
5. **输出**:要它输出什么(SQL 代码 / 文字解释)

### 1.2 基础案例:基础 Prompt

```python
# 自包含 setup
if 'pd' not in globals():
    import numpy as np
    import pandas as pd
    print('Setup OK')
# 演示:Prompt 模板(展示文本,不实际调 API)
prompt_template_1 = '''
你是一名资深数据分析师,擅长写高性能 SQL。

任务:写一个 SQL 查询,统计最近 7 天每天的活跃用户数(DAU)。

表结构:
- user_events 表:user_id, event_type, event_time
- event_type 包含 'pv', 'cart', 'purchase'
- 活跃用户定义:当天有任意 event 的用户

约束:
- 用窗口函数补齐缺失日(没数据的天也要输出,DAU=0)
- 加 7 日移动平均
- 按 DAU 降序

输出:可执行的 SQL 代码 + 简要解释
'''
print('Prompt 模板 1(写 SQL):')
print(prompt_template_1)
```

### 1.4 面试题

- Q1: Prompt 怎么写更有效?
  A: 5 要素全(角色/任务/上下文/约束/输出),别让 AI 猜。给具体的表结构、具体业务。
- Q2: AI 写 SQL 怎么保证对?
  A: 1) 自己 review 逻辑;2) 跑通测试;3) 看 EXPLAIN 性能;4) 边界条件手动验。
- Q3: AI 写出来的代码出现幻觉怎么办?
  A: 1) 怀疑态度,所有代码都要 review;2) 用 lint 工具查语法;3) 跑测试用例。
